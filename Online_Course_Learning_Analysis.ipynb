# Online Course and Learning Platform Analysis

This notebook explores a **synthetic sample dataset** of online courses. It demonstrates data loading, cleaning, exploratory analysis, and visualizations.

> Important: The included dataset is artificially generated for learning and project demonstration. It is not real platform data.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../dataset/online_courses.csv')
df.head()
```

```python
print("Dataset shape:", df.shape)
display(df.info())
display(df.describe(numeric_only=True))
print("Missing values:\n", df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Clean duplicate rows and validate values
df = df.drop_duplicates()
df = df[df['Course_Rating'].between(0, 5)]
df = df[df['Completion_Rate_Percent'].between(0, 100)]
df = df[df['Students_Enrolled'] >= 0]
print("Cleaned dataset shape:", df.shape)
```

```python
# 1. Top 10 courses by enrolment
top_courses = (df.groupby('Course_Name', as_index=False)['Students_Enrolled']
               .sum().nlargest(10, 'Students_Enrolled')
               .sort_values('Students_Enrolled'))
plt.figure(figsize=(10, 6))
plt.barh(top_courses['Course_Name'], top_courses['Students_Enrolled'])
plt.title('Top 10 Courses by Total Enrolment')
plt.xlabel('Students Enrolled')
plt.tight_layout()
plt.savefig('../visualizations/top_courses.png', dpi=150)
plt.show()
```

```python
# 2. Average rating by category
ratings = df.groupby('Category')['Course_Rating'].mean().sort_values()
plt.figure(figsize=(9, 5))
plt.barh(ratings.index, ratings.values)
plt.title('Average Course Rating by Category')
plt.xlabel('Average Rating (out of 5)')
plt.xlim(0, 5)
plt.tight_layout()
plt.savefig('../visualizations/category_ratings.png', dpi=150)
plt.show()
```

```python
# 3. Average completion rate by difficulty
completion = df.groupby('Difficulty_Level')['Completion_Rate_Percent'].mean().reindex(
    ['Beginner', 'Intermediate', 'Advanced'])
plt.figure(figsize=(7, 4))
plt.bar(completion.index, completion.values)
plt.title('Average Completion Rate by Difficulty')
plt.ylabel('Completion Rate (%)')
plt.ylim(0, 100)
plt.tight_layout()
plt.savefig('../visualizations/completion_by_difficulty.png', dpi=150)
plt.show()
```

```python
# 4. Enrolments by platform
platform_enrolments = df.groupby('Platform')['Students_Enrolled'].sum().sort_values()
plt.figure(figsize=(8, 5))
plt.barh(platform_enrolments.index, platform_enrolments.values)
plt.title('Total Enrolments by Platform')
plt.xlabel('Students Enrolled')
plt.tight_layout()
plt.savefig('../visualizations/platform_enrolments.png', dpi=150)
plt.show()
```

```python
# 5. Course price vs rating
plt.figure(figsize=(8, 5))
plt.scatter(df['Course_Price_USD'], df['Course_Rating'], alpha=0.6)
plt.title('Course Price vs Rating')
plt.xlabel('Course Price (USD)')
plt.ylabel('Course Rating')
plt.tight_layout()
plt.savefig('../visualizations/price_vs_rating.png', dpi=150)
plt.show()
```

```python
# Summary tables to support your written findings
display(df.groupby('Category').agg(
    Courses=('Course_ID', 'count'),
    Average_Rating=('Course_Rating', 'mean'),
    Average_Completion=('Completion_Rate_Percent', 'mean'),
    Total_Enrolments=('Students_Enrolled', 'sum')
).round(2).sort_values('Total_Enrolments', ascending=False))
```
